```python
#j)	Split Date added into 3 separate columns having date value in 1st column, month value in 2nd column and year value in 3rd.     
netflix_df['date_added'] = pd.to_datetime(netflix_df['date_added'])
# Extract day, month, and year into new columns
netflix_df['day_added'] = netflix_df['date_added'].dt.day
netflix_df['month_added'] = netflix_df['date_added'].dt.month_name() 
netflix_df['year_added'] = netflix_df['date_added'].dt.year
netflix_df.head()
```

```python
student_df= pd.read_csv("C:\\Users\\richa\\student.csv") 
```

```python
marks_df= pd.read_csv("C:\\Users\\richa\\marks.csv")
```

```python
combined_df = pd.concat([student_df, marks_df],axis=0) #a) Combine both the dataframes into single dataframe which will have all the records from both the tables.
combined_df
```

```python
combined_df=pd.merge(student_df,marks_df,how="outer",on="Student_id")
combined_df
```

```python
print("Maximum marks Gender-wise:")  #b)	Print the maximum marks Gender wise
combined_df.groupby('Gender')['Mark'].agg(['max'])
```

```python
print("Minimum marks Gender-wise:")  #b)	Print theminimum marks Gender wise
combined_df.groupby('Gender')['Mark'].agg(['min']) 
```

```python
#c)	Print all the students IDs and their marks who have scored more than the average marks of the class.
average_marks = combined_df['Mark'].mean()
students_above_average = combined_df[combined_df['Mark'] > average_marks]
students_above_average[['Student_id','Mark']]
```

```python
males_employed_df = combined_df[(combined_df['Gender'] == 'Male') & (combined_df['Employed'] == 'yes')] #d)	Print the dataframe who are Males and are Employed.
males_employed_df
```

```python
#e)	Create a new Column ‘IQ_level’ which will have 3 values (Intelligent, Mediocre, weak). If student scored than 80 then Tag him as Intelligent, if student scored between 50-80, then Mediocre, else weak
def score(Mark):
    if Mark>80:
        return "Intelligent"
    elif Mark >50 and Mark <80:
        return "Mediocre"
    else:
        return "weak"
```

```python
combined_df["IQ_level"]=combined_df["Mark"].apply(score)
combined_df.sample(10)
```

```python
combined_df.groupby('City')['Gender'].value_counts()  #f)	Count the number of males and females from each city.
```

```python
male_students=combined_df[combined_df['Gender'] == 'Male'] #g)	Print the top 5 Male scorers
top_5_male_scorers=male_students.sort_values(by='Mark',ascending=False).head(5)
top_5_male_scorers
```
